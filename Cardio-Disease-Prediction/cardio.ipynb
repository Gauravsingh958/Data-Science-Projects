```python
# dependencies
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import PowerTransformer
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import RobustScaler , StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score , precision_score, recall_score, confusion_matrix, ConfusionMatrixDisplay
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

```

```python
def power_transform(numerical):
    
    pt = PowerTransformer()
    
    num_cardio = pt.fit_transform(numerical)
    
    return num_cardio
```

```python
num_cardio = cardio[numerical]
num_cardio = power_transform(num_cardio)
```

## Splitting Data :

```python
x= cardio.drop("cardio", axis = 1)
y= cardio["cardio"]
```

```python
x_train,x_test,y_train,y_test= train_test_split(x,y, train_size= 0.75)
print(f" x_train shape is : {x_train.shape}\n x_test shape is : {x_test.shape}")
```

## Scaling The Numerical Columns :

```python
numerical
```

```python
numerical_ct = ColumnTransformer(transformers=[('ROBSC', RobustScaler(),numerical)],
                               
                               remainder= "passthrough")
```

```python
numerical_ct
```

```python
x_train_transfomed = numerical_ct.fit_transform(x_train)
print(x_train_transfomed)
```

```python
numerical_ct.get_feature_names_out()
```

```python
pd.DataFrame(data= x_train_transfomed, columns=numerical_ct.get_feature_names_out() )
```

```python
x_test_transformed = numerical_ct.fit_transform(x_test)
```

```python
cardio
```

## Model Training

```python
# DEcisicion Tree Classifier

Dtc = DecisionTreeClassifier()
Dtc.fit(x_train_transfomed, y_train)
dtc_pred= Dtc.predict(x_test_transformed)
print(dtc_pred)
```

```python
# Metrics
model_acc = accuracy_score(y_test, dtc_pred)
rec = recall_score(y_test, dtc_pred)
pres = precision_score(y_test, dtc_pred)
cm = confusion_matrix(y_test,dtc_pred)
print(model_acc)
print(rec)
print(pres)
```

```python
ConfusionMatrixDisplay(cm).plot()
```

```python
# with params

dtc1 = DecisionTreeClassifier(max_depth=5, min_samples_leaf=3, criterion= 'entropy')
dtc1.fit(x_train_transfomed,y_train)
prids=  dtc1.predict(x_test_transformed)
cm1 = confusion_matrix(y_test,prids)
```

```python
acc1 = accuracy_score(prids, y_test)
print(acc1)
```

```python
ConfusionMatrixDisplay(cm1).plot()
```

## Hyperparameter tuning :

```python
Dtc.get_params()
```

```python
param_dist = {
    "max_depth":[4,5,6,7,8],
    "min_samples_split":[2,5,10],
    "min_samples_leaf":[1,2,3,5],
    "criterion":["gini","entropy"]
}


random_cv = RandomizedSearchCV( Dtc,
    param_distributions=param_dist,
    n_iter=20,
    cv=5,)

random_cv.fit(x_train_transfomed, y_train)
```

```python
random_cv.best_params_
```

```python
preds = random_cv.predict(x_test_transformed)
```

```python
acc = accuracy_score(preds, y_test)
rec = recall_score(y_test, preds)
pres = precision_score(y_test, preds)

print(f"Accuracy_score = {acc}")
print(f"recall_score = {rec}")
print(f"precision_score = {pres}")
```

```python
rfc = RandomForestClassifier(criterion= 'entropy', n_estimators=150,
                              min_samples_leaf= 5, max_depth= 6)
rfc.fit(x_train_transfomed, y_train)
rfc_preds = rfc.predict(x_test_transformed)
```

```python
accuracy_score(y_test, rfc_preds)
```

```python
rfc.get_params() 
```

```python
rfc_params= {'criterion': ['gini', 'entropy'],
 'max_depth':[ 6,4,7,8],
 'max_samples': [3,5,6],
 'min_samples_leaf': [2,4,6,8],
 'min_samples_split': [2,4,5],
 'n_estimators': [150,200,300]
}
rfc_tune  = RandomizedSearchCV( Dtc,
    param_distributions=param_dist,
    n_iter=20,
    cv=5,)

rfc_tune.fit(x_train_transfomed, y_train)
```

```python
rfc_tune.best_params_
```

```python
preds = rfc_tune.predict(x_test_transformed)
```

```python
accuracy_score(y_test, preds)
```

```python
xgb = XGBClassifier()
xgb.fit(x_train_transfomed, y_train)
xgb_preds =  xgb.predict(x_test_transformed)

acc_score = accuracy_score(y_test,xgb_preds)
rec_score = recall_score(y_test,xgb_preds)
presc = precision_score(y_test,xgb_preds)

print(acc_score)
print(rec_score)
print(presc)

cm = confusion_matrix(y_test,xgb_preds)
```
